## Modelo aumentado y discretización

Para $b$ constante, o aproximadamente constante durante una muestra,
$$
\dot z=\begin{bmatrix}v\\ (u-bv)/m\\0\end{bmatrix},\qquad y=[1\;0\;0]z+\nu.
$$
Aunque la planta es lineal si $b$ se conoce, el modelo aumentado es **no lineal** por el producto $bv$.

El Euler de los adjuntos es
$$
f_E(z,u)=\begin{bmatrix}p+T_sv\\v+T_s(u-bv)/m\\b\end{bmatrix},\quad
F_E=\begin{bmatrix}1&T_s&0\\0&1-T_sb/m&-T_sv/m\\0&0&1\end{bmatrix}.
$$
Esto sirve para entender el EKF, pero un certificado de Euler no es automáticamente un certificado de la planta física muestreada.

Aquí mantendremos $u_k$ y $b_k$ constantes en cada intervalo. Definiendo
$$
a=e^{-bT_s/m},\qquad g=\frac{m(1-a)}b,\qquad h=\frac{T_s-g}b,
$$
obtenemos la discretización ZOH exacta:
$$
s_{k+1}=A_d(b_k)s_k+B_d(b_k)u_k+w_k,\quad
A_d=\begin{bmatrix}1&g\\0&a\end{bmatrix},\quad B_d=\begin{bmatrix}h\\g/m\end{bmatrix}.
$$
La fricción sigue un paseo aleatorio **en el filtro**, $b_{k+1}=b_k+w_{b,k}$. Es una forma de permitir adaptación, no una afirmación de que la fricción real sea gaussiana. Las covarianzas elegidas son discretas; al modificar $T_s$ no deben reutilizarse sin revisar su escala.

```python
def zoh_coeff(b, dt=DT, mass=MASS):
    # El EKF se mantiene alejado de b=0; expm1 evita cancelación en 1-exp(-x).
    a = np.exp(-b*dt/mass)
    g = -mass*np.expm1(-b*dt/mass)/b
    h = (dt-g)/b
    return a, g, h

def model(b, dt=DT, mass=MASS):
    a, g, h = zoh_coeff(b, dt, mass)
    return np.array([[1.0, g], [0.0, a]]), np.array([[h], [g/mass]])

def f_aug(z, u):
    A, B = model(z[2])
    return np.r_[A @ z[:2] + B[:, 0]*u, z[2]]

def jac_aug(z, u):
    _, vel, b = z
    a, g, h = zoh_coeff(b)
    da = -DT*a/MASS
    dg = DT*a/b - MASS*(1.0-a)/b**2
    dh = -dg/b - (DT-g)/b**2
    return np.array([[1.0, g, dg*vel + dh*u],
                     [0.0, a, da*vel + dg*u/MASS],
                     [0.0, 0.0, 1.0]])

# Dos comprobaciones que sí detectan errores de modelado.
for b_test in [0.05, 0.5, 2.0, 5.0, 20.0]:
    Ac = np.array([[0., 1.], [0., -b_test/MASS]])
    Bc = np.array([[0.], [1./MASS]])
    exact = expm(np.block([[Ac, Bc], [np.zeros((1, 3))]])*DT)
    A, B = model(b_test)
    assert np.allclose(np.c_[A, B], exact[:2], atol=1e-11)
    z_test, u_test, eps = np.array([1., 0.7, b_test]), 2.3, 1e-6
    J_fd = np.column_stack([(f_aug(z_test+eps*np.eye(3)[i], u_test)
                            -f_aug(z_test-eps*np.eye(3)[i], u_test))/(2*eps)
                           for i in range(3)])
    assert np.allclose(jac_aug(z_test, u_test), J_fd, atol=2e-7)
print('ZOH y Jacobiano comprobados contra exponencial matricial y diferencias finitas.')
```

## ¿Se puede estimar la fricción midiendo sólo posición?

Con una entrada conocida, las primeras derivadas de la salida son
$$
y=p,\quad\dot y=v,\quad\ddot y=(u-bv)/m.
$$
La matriz de sensibilidades respecto a $[p,v,b]$ es
$$
\mathcal O_{\rm local}=
\begin{bmatrix}1&0&0\\0&1&0\\0&-b/m&-v/m\end{bmatrix},\qquad
\det\mathcal O_{\rm local}=-v/m.
$$
Si $v\ne0$, este test da rango completo. Si el vehículo permanece en reposo con $u=0$, cualquier $b$ produce la misma salida: no hay información para identificarlo. Que el determinante sea cero en **un instante** no implica imposibilidad permanente; una aceleración posterior puede aportar información.

Para estimación mantenida interesa la excitación a lo largo del tiempo; en este modelo, $\int_t^{t+T}v^2(\tau)\,d\tau$ es un indicador útil. No equivale por sí solo a un teorema completo de convergencia del EKF con posición ruidosa. Tampoco se deben calcular dos derivadas de la medición como sustituto del filtro: amplificarían mucho el ruido.

## EKF conjunto: predicción, corrección y diagnóstico

Desde el posterior $(\hat z_{k|k},P_{k|k})$, después de aplicar $u_k$:
$$
\hat z^-_{k+1}=f(\hat z_{k|k},u_k),\quad
P^-_{k+1}=F_kP_{k|k}F_k^T+Q_{\rm EKF},
$$
$$
\epsilon_{k+1}=y_{k+1}-H\hat z^-_{k+1},\quad
\Omega_{k+1}=HP^-_{k+1}H^T+R_y,\quad
L_{k+1}=P^-_{k+1}H^T\Omega_{k+1}^{-1},
$$
$$
\hat z_{k+1|k+1}=\hat z^-_{k+1}+L_{k+1}\epsilon_{k+1}.
$$
Usaremos la actualización de Joseph:
$$
P^+=(I-LH)P^-(I-LH)^T+LR_yL^T.
$$
Reduce problemas numéricos de simetría/positividad; no arregla una mala linealización o falta de observabilidad.

Recortamos únicamente $\hat b$ al intervalo amplio `B_EST`. Esto evita valores físicos absurdos y singularidades numéricas, pero no convierte al EKF en un filtro gaussiano exacto condicionado a restricciones. Las bandas $\hat b\pm2\sqrt{P_{bb}}$ serán **diagnósticas aproximadas**, no cotas duras para la estabilidad. El NIS es $\epsilon^2/\Omega$; esperar media cercana a 1 sólo tiene sentido si las hipótesis y la sintonización son razonables.

```python
def ekf_step(zhat, P, u, y_next):
    F = jac_aug(zhat, u)                  # posterior anterior, no estado predicho
    zminus = f_aug(zhat, u)
    Pminus = F @ P @ F.T + Q_EKF
    innovation = float(y_next-zminus[0])
    omega = float(Pminus[0, 0]+R_Y)
    gain = Pminus[:, 0]/omega
    znew = zminus + gain*innovation
    raw_b = znew[2]
    znew[2] = np.clip(raw_b, *B_EST)
    J = np.eye(3)-np.outer(gain, H[0])
    Pnew = J @ Pminus @ J.T + R_Y*np.outer(gain, gain)
    Pnew = (Pnew+Pnew.T)/2
    return znew, Pnew, innovation**2/omega, raw_b != znew[2]
```

## Propuesta 1: asignación de polos adaptativa (“naive”)

Para una referencia constante $r$, define $e=p-r$, y aplica
$$
u=-k_p(\hat p-r)-k_v\hat v.
$$
Si los estados y la fricción se conocieran exactamente, el polinomio continuo sería
$$
s^2+\frac{b+k_v}{m}s+\frac{k_p}{m}.
$$
Igualándolo a $(s+p_1)(s+p_2)$, con $p_1,p_2>0$, se obtiene la regla correcta del PDF:
$$
\boxed{k_p=mp_1p_2,\qquad k_v=m(p_1+p_2)-\hat b.}
$$
La fricción real aparece en el amortiguamiento efectivo como
$$
b+k_v=m(p_1+p_2)+(b-\hat b).
$$
Una sobreestimación suficientemente grande puede hacerlo negativo. Para $m=1,p_1=2,p_2=3,b=2$, basta $\hat b>7$. No hace falta una conmutación complicada para perder estabilidad en ese caso.

```python
def gain_naive(bhat, poles=(2.0, 3.0)):
    p1, p2 = poles
    return np.array([MASS*p1*p2, MASS*(p1+p2)-bhat], dtype=float)

for bhat in [2.0, 9.0]:
    K = gain_naive(bhat)
    F_cont = np.array([[0., 1.], [-K[0]/MASS, -(2.0+K[1])/MASS]])
    print(f'b real=2, b estimada={bhat:g}: K={K}, polos reales={np.linalg.eigvals(F_cont)}')
```

## Propuesta 2: EKF + LQR adaptativo

El LQR discreto usa $e_k=s_k-[r,0]^T$ y el costo determinista
$$
J=\sum_{k=0}^{\infty}(e_k^TQ_ce_k+u_k^TR_uu_k).
$$
Para un modelo fijo, se resuelve
$$
S=A_d^TSA_d-A_d^TSB_d(R_u+B_d^TSB_d)^{-1}B_d^TSA_d+Q_c,
$$
$$
K=(R_u+B_d^TSB_d)^{-1}B_d^TSA_d,\qquad u_k=-K\hat e_k.
$$
Podemos sustituir $b$ por $\hat b_k$ y se resuelve Riccati en cada muestra. Haremos esto, midiendo sólo posición. El optimizador elige un compromiso entre seguimiento y esfuerzo para el **modelo congelado**, no el controlador adaptativo globalmente óptimo.

El LQG clásico combina LQR y KF para un modelo lineal conocido, con hipótesis adecuadas de ruido, costo y estabilizabilidad/detectabilidad. La separación explica cómo diseñar estimación y control. Hay también formulaciones LQG de tiempo variable conocido; recalcular una DARE de horizonte infinito usando un parámetro desconocido estimado en lazo cerrado no es esa formulación. En nuestro caso, el estado aumentado es no lineal, el EKF es aproximado y la entrada afecta la información disponible sobre $b$. No se heredan automáticamente optimalidad ni estabilidad [1].

```python
def gain_lqr(bhat, q=Q_COST, ru=R_COST):
    A, B = model(float(bhat))
    S = solve_discrete_are(A, B, q, np.array([[ru]]))
    return np.linalg.solve(np.array([[ru]])+B.T @ S @ B,
                           B.T @ S @ A).ravel()

b_grid = np.linspace(*B_PHYS, 100)
gains = np.array([gain_lqr(b) for b in b_grid])
fig, ax = plt.subplots(1, 2, figsize=(10.5, 3.4))
ax[0].plot(b_grid, gains[:, 0]); ax[0].set_ylabel('k_p [N/m]')
ax[1].plot(b_grid, gains[:, 1]); ax[1].set_ylabel('k_v [N·s/m]')
for a in ax: a.set_xlabel('Fricción usada para diseñar [N·s/m]')
fig.suptitle('Ganancias del LQR discreto con pesos fijos'); fig.tight_layout(); plt.show()
```

## Banco de simulación común

En cada muestra, las propuestas 1–2 usan el posterior disponible del EKF para decidir $u_k$. La planta avanza con ZOH, se mide **sólo posición**, y se hace la nueva predicción/corrección. El estado verdadero aparece únicamente en la simulación y en las métricas; jamás entra al controlador ni al filtro.

Comparamos con la misma semilla, condiciones iniciales, ruido y covarianzas. El EKF comienza con $\hat b_0=1$, mientras que $b=2$. Para referencias móviles añadimos
$$
u_{\rm ff}=m\ddot r+\hat b\dot r,\quad
u=u_{\rm ff}-K([\hat p,\hat v]^T-[r,\dot r]^T).
$$
Esta compensación es aproximada con $\hat b\ne b$ y muestreo. Una referencia móvil no es un equilibrio: la garantía rigurosa se formulará como seguimiento con perturbaciones, no como convergencia exacta prometida para toda trayectoria.

```python
def reference(t, mode):
    if mode == 'setpoint':
        return 5.0, 0.0, 0.0
    if mode == 'moving':
        return 2.0+1.2*np.sin(0.7*t), 0.84*np.cos(0.7*t), -0.588*np.sin(0.7*t)
    raise ValueError(mode)

def friction(t, mode):
    if mode == 'constant':
        return 2.0
    return 2.0 if t < 8.0 else (4.2 if t < 18.0 else 1.0)

def simulate(method, duration=24., ref_mode='setpoint', b_mode='constant',
             seed=7, noisy=True, bhat0=1.0):
    rng = np.random.default_rng(seed)
    n = int(round(duration/DT))
    t = DT*np.arange(n+1)
    x = np.zeros((n+1, 2))
    z = np.zeros((n+1, 3)); z[0] = [0.15, 0.0, bhat0]
    obs = np.zeros((n+1, 2)); obs[0] = z[0, :2]
    cov = np.zeros((n+1, 3, 3)); cov[0] = P_INITIAL.copy()
    y = np.empty(n+1); y[0] = x[0, 0] + noisy*SIGMA_Y*rng.normal()
    u = np.empty(n); Khist = np.empty((n, 2)); nis = np.empty(n)
    theta = np.ones(n); margins = np.full(n, np.nan); clips = np.zeros(n, bool)
    refs = np.array([reference(ti, ref_mode) for ti in t])
    bs = np.array([friction(ti, b_mode) for ti in t])
    for k in range(n):
        if method == 'naive':
            K = gain_naive(z[k, 2]); feedback = z[k, :2]
        elif method == 'lqr':
            K = gain_lqr(z[k, 2]); feedback = z[k, :2]
        elif method == 'safe':
            K, theta[k], margins[k] = filter_gain(gain_lqr(z[k, 2]))
            feedback = obs[k]      # observador certificado, separado del EKF
        else:
            raise ValueError(method)
        rr, vr, ar = refs[k]
        bff = np.clip(z[k, 2], *B_PHYS)
        u[k] = MASS*ar + bff*vr - K @ (feedback-refs[k, :2])
        Khist[k] = K
        if method == 'safe':
            obs[k+1] = A0@obs[k] + B0[:, 0]*u[k] + L_OBS[:, 0]*(y[k]-obs[k, 0])
        A, B = model(bs[k])
        x[k+1] = A@x[k] + B[:, 0]*u[k] + noisy*np.sqrt(np.diag(Q_TRUE))*rng.normal(size=2)
        y[k+1] = x[k+1, 0]+noisy*SIGMA_Y*rng.normal()
        z[k+1], cov[k+1], nis[k], clips[k] = ekf_step(z[k], cov[k], u[k], y[k+1])
    return dict(t=t, x=x, z=z, obs=obs, P=cov, y=y, u=u, K=Khist,
                nis=nis, b=bs, ref=refs, theta=theta, margin=margins,
                clipped=clips, method=method)

NAMES = {'naive': 'Polos + EKF', 'lqr': 'LQR + EKF', 'safe': 'LQR filtrado + observador'}

def metrics(r):
    e = r['x'][:, 0]-r['ref'][:, 0]
    mask = r['t'] >= 4.0
    stage = np.einsum('ni,ij,nj->n', r['x'][:-1]-r['ref'][:-1, :2], Q_COST,
                      r['x'][:-1]-r['ref'][:-1, :2])+R_COST*r['u']**2
    return {'Método': NAMES[r['method']], 'RMSE p [m]': np.sqrt(np.mean(e**2)),
            'RMSE p desde 4 s [m]': np.sqrt(np.mean(e[mask]**2)),
            'RMSE b': np.sqrt(np.mean((r['z'][mask, 2]-r['b'][mask])**2)),
            'Esfuerzo ∫u²dt': DT*np.sum(r['u']**2), 'Pico |u| [N]': np.max(abs(r['u'])),
            'Costo finito ×dt': DT*np.sum(stage), 'Recortes b': int(r['clipped'].sum())}

def plot_comparison(results, title):
    fig, ax = plt.subplots(3, 2, figsize=(12, 10), sharex=True)
    first = results[0]
    ax[0, 0].plot(first['t'], first['ref'][:, 0], 'k--', label='Referencia')
    ax[1, 0].plot(first['t'], first['b'], 'k--', label='Fricción real')
    for rr in results:
        t, label = rr['t'], NAMES[rr['method']]
        ax[0, 0].plot(t, rr['x'][:, 0], label=label)
        ax[0, 1].plot(t, rr['x'][:, 0]-rr['ref'][:, 0], label=label)
        ax[1, 0].plot(t, rr['z'][:, 2], label=label)
        ax[1, 1].plot(t[:-1], rr['u'], label=label)
        ax[2, 0].plot(t[:-1], rr['K'][:, 0], label=label)
        ax[2, 1].plot(t[:-1], rr['K'][:, 1], label=label)
    for a, lab in zip(ax.ravel(), ['Posición [m]', 'Error p-r [m]', 'Fricción [N·s/m]',
                                  'Fuerza [N]', 'k_p [N/m]', 'k_v [N·s/m]']):
        a.set_ylabel(lab)
    ax[0, 0].legend(fontsize=8); ax[1, 0].legend(fontsize=8)
    for a in ax[-1]: a.set_xlabel('Tiempo [s]')
    fig.suptitle(title); fig.tight_layout(); plt.show()
```

```python
reg_naive = simulate('naive', b_mode='moving', ref_mode='moving')
reg_lqr = simulate('lqr', b_mode='moving', ref_mode='moving')
plot_comparison([reg_naive, reg_lqr], 'Experimento A · regulación con b=2 y sólo posición medida')
display(pd.DataFrame([metrics(reg_naive), metrics(reg_lqr)]).set_index('Método').round(4))
```

## Por qué recalcular ganancias no basta

### Tres problemas diferentes

1. **Modelo equivocado.** $A_d(\hat b)-B_d(\hat b)K$ puede ser estable y el lazo con $b$ real no serlo. El ejemplo $\hat b=9,b=2$ de la propuesta 1 ya lo muestra.
2. **Cambio de dinámica.** Para $e_{k+1}=F_ke_k$, la solución depende del producto $F_{k-1}\cdots F_1F_0$. Tener $\rho(F_k)<1$ en cada muestra no basta para que ese producto decaiga.
3. **Estimación y control acoplados.** Con $\tilde s=s-\hat s$, el error de estimación entra como una perturbación $B_dK\tilde s$. El EKF aumentado no tiene, en general, una garantía global de convergencia; además, controlar modifica la excitación que permite estimar $b$.

Para ver el punto 2 mediante Riccati, si cada $S_k$ es una matriz de Lyapunov del modelo congelado, sólo sabes algo sobre
$F_k^TS_kF_k-S_k$. Pero para $V_k=e_k^TS_ke_k$, el cambio real es
$$
V_{k+1}-V_k=e_k^T\big(F_k^TS_{k+1}F_k-S_k\big)e_k.
$$
Falta controlar el término $F_k^T(S_{k+1}-S_k)F_k$. En continuo aparece el análogo $\dot S$. Una gráfica de polos instantáneos no contiene esa información [3, 4].

### Contraejemplo dentro de la forma mecánica de segundo orden

Considera $m=1,b=0.1,k_v=0.1$ y alterna $k_p=1$ con $k_p=100$. Los dos sistemas
$$
F_1=\begin{bmatrix}0&1\\-1&-0.2\end{bmatrix},\qquad
F_2=\begin{bmatrix}0&1\\-100&-0.2\end{bmatrix}
$$
tienen polos con parte real $-0.1$. Sin embargo, la conmutación siguiente produce crecimiento. Es un ejemplo construido para refutar la regla general de los polos congelados; **no afirma que el LQR con los pesos anteriores genere esta secuencia**.

```python
F1 = np.array([[0., 1.], [-1., -0.2]])
F2 = np.array([[0., 1.], [-100., -0.2]])
T1 = np.pi/(2*np.sqrt(1.0-0.1**2))
T2 = np.pi/(2*np.sqrt(100.0-0.1**2))
Phi = expm(F2*T2) @ expm(F1*T1)
print('Polos F1:', np.linalg.eigvals(F1))
print('Polos F2:', np.linalg.eigvals(F2))
print(f'T1={T1:.4f} s, T2={T2:.4f} s; radio espectral del ciclo={max(abs(np.linalg.eigvals(Phi))):.4f}')
assert max(np.linalg.eigvals(F1).real) < 0 and max(np.linalg.eigvals(F2).real) < 0
assert max(abs(np.linalg.eigvals(Phi))) > 1
q = np.array([1., 0.]); tt = [0.]; qq = [q.copy()]
for cycle in range(7):
    for F, dur in [(F1, T1), (F2, T2)]:
        base_t, base_q = tt[-1], q.copy()
        for tau in np.linspace(0, dur, 60)[1:]:
            q = expm(F*tau) @ base_q
            tt.append(base_t+tau); qq.append(q.copy())
fig, ax = plt.subplots(figsize=(10.5, 3.6))
ax.semilogy(tt, np.linalg.norm(qq, axis=1))
ax.set(xlabel='Tiempo [s]', ylabel='Norma del estado',
       title='Dos lazos congelados estables; su alternancia es inestable')
plt.show()
```

La transición de un ciclo es
$$
e(T_1)=e^{F_1T_1}e(0),\qquad
e(T_1+T_2)=\underbrace{e^{F_2T_2}e^{F_1T_1}}_{\Phi}e(0).
$$
La matriz que actúa primero queda a la derecha. Después de $n$ ciclos, $e_n=\Phi^ne_0$. Si $\rho(\Phi)>1$, existe una dirección que crece. El cálculo utiliza exponenciales exactas, por lo que este crecimiento no es un artefacto de integración numérica.

### Un matiz crucial para este modelo: no toda adaptación es peligrosa

Si $k_p>0$ es **constante**, los estados son exactos y
$0<d_{\min}\le b(t)+k_v(t)\le d_{\max}<\infty$, la energía
$$
V=\tfrac12 k_pe^2+\tfrac12 mv^2,\qquad
\dot V=-(b+k_v)v^2\le0
$$
ya evita crecimiento energético. Con las cotas uniformes anteriores se puede obtener una función estricta añadiendo un término cruzado pequeño $\varepsilon ev$, y demostrar decaimiento uniforme. No hace falta limitar la rapidez de $k_v$ porque $V$ no depende de esa ganancia.

En la propuesta 1, recortar $k_v\ge d_{\min}-b_{\min}$ es una mejora sencilla bajo esas hipótesis. Si además se exige $k_v\ge0$ y $b\ge b_{\min}>0$, se conserva amortiguamiento positivo.

Incluso el **LQR continuo** de este modelo, con $Q_c=\operatorname{diag}(q_p,q_v)$ fijo y $R_u>0$, tiene
$$
k_p=\sqrt{q_p/R_u},\qquad
k_v=-\hat b+\sqrt{\hat b^2+2mk_p+q_v/R_u}.
$$
Por tanto, $k_p$ es constante y $k_v>0$. Con fricción real positiva y estados exactos existe esta estructura estabilizante. Sería incorrecto usar el contraejemplo anterior para afirmar que ese LQR continuo necesariamente se vuelve inestable.

Lo que falta para **los programas reales** es cubrir discretización, incertidumbre, estimación y posibles restricciones. La solución final los abordará para un modelo muestreado concreto.

### Tres caminos para una garantía

| Camino | Qué se demuestra | Qué no basta hacer |
|---|---|---|
| Lyapunov común | Un mismo $P_c\succ0$ decrece para todos los lazos admisibles. Permite cambios arbitrarios. | Revisar sólo polos o unos pocos puntos sin justificar la cobertura. |
| Lyapunov dependiente del parámetro | Se acota el término $\partial P/\partial\theta\,\dot\theta$, además de la incertidumbre. | Suavizar $\hat b$ con una constante elegida a ojo. |
| Tiempo de permanencia | Si $\dot V_i\le-\lambda V_i$ y $V_j\le\mu V_i$, basta permanecer más de $\log\mu/\lambda$ por modo, en el caso determinista considerado. | Actualizar cada segundo sin calcular esas cotas. |

Un UKF, una proyección de parámetros o un filtro pasa-bajas pueden mejorar resultados, pero no reemplazan estas demostraciones. Elegimos **Lyapunov común** porque produce un filtro de ganancias pequeño y una garantía que no exige limitar su velocidad de actualización [3, 4].

## Solución final: adaptar dentro de un conjunto certificado

### Por qué añadimos un observador independiente

Un filtro de ganancias que sólo verifica $(A-BK)^TP_c(A-BK)-P_c\prec0$ certifica **realimentación de estado exacto**. Si después se usa un EKF arbitrario, la prueba queda condicionada a su error. Para cerrar esa brecha, usaremos:

- El mismo EKF para estimar $[p,v,b]$ y proponer $K_{\rm LQR}(\hat b_k)$.
- Un observador lineal fijo que utiliza **únicamente la posición medida** para la realimentación.
- Un filtro de ganancias que certifica conjuntamente planta y observador, incluyendo el desajuste de fricción.

Así, el EKF sigue siendo el mecanismo de adaptación, pero su convergencia no es una hipótesis de estabilidad del lazo físico. Sus estimaciones de estado quedan disponibles como diagnóstico. Se pierde la pretensión de optimalidad LQG exacta; se gana una garantía robusta explícita para el modelo indicado.

El observador nominal, con $b_0=2.5$, es
$$
\bar s_{k+1}=A_0\bar s_k+B_0u_k+L_o(y_k-C\bar s_k),\qquad
u_k=-K_k(\bar s_k-s_r),\quad s_r=[r,0]^T.
$$
Es un observador predictor: usa $y_k$ para obtener $\bar s_{k+1}$; no debe intercambiarse su orden con una actualización correctora sin rehacer el certificado.

Para referencia constante, sin ruido, define
$$
q_k=\begin{bmatrix}s_k-s_r\\\bar s_k-s_r\end{bmatrix}.
$$
Entonces
$$
q_{k+1}=\mathcal F(b_k,K_k)q_k,\qquad
\boxed{\mathcal F(b,K)=\begin{bmatrix}
A_d(b)&-B_d(b)K\\
L_oC&A_0-B_0K-L_oC
\end{bmatrix}.}
$$
Esta matriz de dimensión 4 incluye la incertidumbre de la planta; no se invoca un principio de separación adaptativo.

```python
B0_NOMINAL = 2.5
A0, B0 = model(B0_NOMINAL)
L_OBS = place_poles(A0.T, C.T, np.exp(-np.array([8., 10.])*DT)).gain_matrix.T
K_BASE = gain_lqr(B0_NOMINAL)

def augmented_matrix(A, B, K):
    K = np.asarray(K).reshape(1, 2)
    return np.block([[A, -B@K], [L_OBS@C, A0-B0@K-L_OBS@C]])

print('K de respaldo:', K_BASE)
print('L del observador predictor:', L_OBS.ravel())
print('Polos nominales del observador:', np.linalg.eigvals(A0-L_OBS@C))
```

### Cubrir TODO el intervalo de fricción sin confundir una malla con una prueba

La discretización exacta no es afín en $b$; comprobar sólo $b_{\min}$ y $b_{\max}$ no basta. En cambio, sus tres coeficientes son monótonos decrecientes para $b>0$:
$$
a=e^{-bT_s/m},\quad
g=\int_0^{T_s}e^{-b\tau/m}\,d\tau,\quad
h=\frac1m\int_0^{T_s}(T_s-\tau)e^{-b\tau/m}\,d\tau.
$$
En cada subintervalo $[\ell,u]$, los valores reales pertenecen a la caja
$$
[a(u),a(\ell)]\times[g(u),g(\ell)]\times[h(u),h(\ell)].
$$
Sus 8 vértices contienen, por convexidad, todos los modelos del subintervalo. Dividiremos $[0.5,5]$ en seis subintervalos: **48 vértices de cajas**, no 48 muestras de $b$. Las combinaciones independientes incluyen modelos no físicos; eso introduce conservadurismo, pero asegura cobertura.

Para un $K$ fijo, $\mathcal F$ es afín en $(a,g,h)$. Si $P_c\succ0$,
$$
\left(\sum_i\alpha_i\mathcal F_i\right)^TP_c
\left(\sum_i\alpha_i\mathcal F_i\right)
\preceq\sum_i\alpha_i\mathcal F_i^TP_c\mathcal F_i.
$$
Por eso verificar los vértices certifica la caja completa. Esta cobertura analítica es lo que permite pasar de comprobaciones finitas a un intervalo continuo.

```python
def plant_vertices(n_intervals=6):
    edges = np.linspace(*B_PHYS, n_intervals+1)
    vertices = []
    for lo, hi in zip(edges[:-1], edges[1:]):
        bounds = list(zip(zoh_coeff(lo), zoh_coeff(hi)))
        for a, g, h in product(*bounds):
            vertices.append((np.array([[1., g], [0., a]]), np.array([[h], [g/MASS]])))
    return vertices

VERTICES = plant_vertices()
AV = np.array([v[0] for v in VERTICES])
BV = np.array([v[1] for v in VERTICES])

def augmented_vertices(K):
    K = np.asarray(K).reshape(1, 2)
    Fs = np.zeros((len(VERTICES), 4, 4))
    Fs[:, :2, :2] = AV
    Fs[:, :2, 2:] = -BV @ K
    Fs[:, 2:, :2] = L_OBS@C
    Fs[:, 2:, 2:] = A0-B0@K-L_OBS@C
    return Fs

print(f'{len(VERTICES)} vértices cubren todo b ∈ {B_PHYS}.')
```

### Diseño fuera de línea y verificación independiente

Elegimos $\rho_c=0.995<1$. Buscamos un $P_c$ común que certifique la ganancia de respaldo $K_0$:
$$
\max_{P_c,\eta}\eta\quad\text{sujeto a}\quad
P_c\succeq10^{-3}I,\ \operatorname{tr}(P_c)=4,
\quad \rho_cP_c-\mathcal F_i(K_0)^TP_c\mathcal F_i(K_0)\succeq\eta I\quad\forall i.
$$
Es un SDP porque las matrices $\mathcal F_i(K_0)$ ya están fijadas. **No** estamos resolviendo conjuntamente por $K_0,L_o,P_c$: eso introduciría productos bilineales y dejaría de ser este problema convexo.

El solver propone la matriz; después NumPy verifica sus autovalores y todos los residuos. Si no hay margen positivo, detenemos la ejecución y rediseñamos. La existencia de una solución para estos parámetros no garantiza factibilidad para cualquier vehículo o intervalo.

```python
RHO_C = 0.995
Pvar = cp.Variable((4, 4), symmetric=True)
eta = cp.Variable()
Fbase = augmented_vertices(K_BASE)
constraints = [Pvar >> 1e-3*np.eye(4), cp.trace(Pvar) == 4.0]
constraints += [RHO_C*Pvar-F.T@Pvar@F >> eta*np.eye(4) for F in Fbase]
problem = cp.Problem(cp.Maximize(eta), constraints)
problem.solve(solver='CLARABEL')
if problem.status not in ('optimal', 'optimal_inaccurate') or Pvar.value is None:
    raise RuntimeError('No se obtuvo un candidato a certificado: revisar diseño y solver.')
P_CERT = (Pvar.value+Pvar.value.T)/2

def certificate_margin(K):
    Fs = augmented_vertices(K)
    residues = RHO_C*P_CERT-Fs.transpose(0, 2, 1)@P_CERT@Fs
    return float(np.linalg.eigvalsh(residues).min())

MARGIN_BASE = certificate_margin(K_BASE)
TOL_CERT = 1e-7              # umbral positivo, lejos de cero numérico
assert np.linalg.eigvalsh(P_CERT).min() > 1e-4
assert MARGIN_BASE > 100*TOL_CERT, 'El respaldo no quedó certificado con margen suficiente.'
print('Estado del solver:', problem.status)
print('P común =', P_CERT, sep=chr(10))
print('Autovalores de P:', np.linalg.eigvalsh(P_CERT))
print(f'Peor margen verificado del respaldo: {MARGIN_BASE:.8f}')
```

### Filtro de ganancias en línea

Define el conjunto
$$
\mathcal K_{\rm segura}=\{K:\mathcal F_i(K)^TP_c\mathcal F_i(K)\preceq\rho_cP_c\ \forall i\}.
$$
Es convexo para $P_c$ fijo. La condición también puede escribirse como LMI mediante Schur:
$$
\begin{bmatrix}\rho_cP_c&\mathcal F_i(K)^TP_c\\P_c\mathcal F_i(K)&P_c\end{bmatrix}\succeq0.
$$
Podríamos minimizar $\|K-K_{\rm LQR}\|^2$ sujeto a esas LMIs. Para que el ejemplo no necesite un SDP en cada muestra, buscamos sobre el segmento
$$
K(\theta)=K_0+\theta(K_{\rm LQR}-K_0),\qquad0\le\theta\le1.
$$
Si el candidato pasa, $\theta=1$. Si no, bisección encuentra el mayor $\theta$ admisible sobre ese segmento. La factibilidad de $\theta=0$ está garantizada por el diseño previo. **No es la proyección euclídea global** sobre $\mathcal K_{\rm segura}$.

Conservamos siempre el mismo $P_c$, observador, intervalo y período de muestreo. Recalcular un $P_c$ distinto en cada paso sin controlar sus cambios volvería a abrir el problema original.

```python
def filter_gain(K_candidate):
    K_candidate = np.asarray(K_candidate, float).ravel()
    if K_candidate.shape != (2,) or not np.all(np.isfinite(K_candidate)):
        return K_BASE.copy(), 0.0, MARGIN_BASE
    mc = certificate_margin(K_candidate)
    if mc >= TOL_CERT:
        return K_candidate.copy(), 1.0, mc
    lo, hi = 0.0, 1.0
    for _ in range(24):
        theta = (lo+hi)/2
        trial = K_BASE+theta*(K_candidate-K_BASE)
        if certificate_margin(trial) >= TOL_CERT:
            lo = theta
        else:
            hi = theta
    K = K_BASE+lo*(K_candidate-K_BASE)
    margin = certificate_margin(K)
    if margin < TOL_CERT:             # respaldo conservador ante redondeo
        return K_BASE.copy(), 0.0, MARGIN_BASE
    return K, lo, margin

for request in [gain_lqr(2.), gain_lqr(20.), np.array([80., -5.]), np.array([np.nan, 1.])]:
    applied, theta, margin = filter_gain(request)
    print(f'Solicitada {request} → aplicada {applied}; θ={theta:.4f}; margen={margin:.3g}')
```

### Qué garantiza exactamente

Para $V(q)=q^TP_cq$, el filtro asegura
$$
V(q_{k+1})\le\rho_cV(q_k),\qquad
\|q_k\|\le\sqrt{\kappa(P_c)}\,\rho_c^{k/2}\|q_0\|.
$$
Así, **la planta y el observador convergen exponencialmente a una referencia constante**, para cualquier secuencia de ganancias aceptadas y cualquier secuencia $b_k\in[0.5,5]$, bajo estas hipótesis:

1. Masa y muestreo coinciden con el diseño; $b$ y $u$ son constantes dentro de cada muestra.
2. La planta es el modelo ZOH considerado; no hay retardo, saturación ni dinámica omitida.
3. Para la afirmación de convergencia exacta, no hay ruido ni perturbaciones.
4. Se usa el observador y el orden de actualización certificados.

No hace falta $\hat b_k\to b_k$, ni una cota de $\dot{\hat b}$, ni polos congelados del EKF. El algoritmo debe poder producir una propuesta finita o activar el respaldo. **La prueba no certifica convergencia ni acotación de la covarianza interna del EKF**; certifica el lazo físico y el observador que se realimenta.

Con ruido o referencia móvil,
$$
q_{k+1}=\mathcal F_kq_k+d_k.
$$
La norma $\|q\|_{P_c}=\sqrt{q^TP_cq}$ satisface
$$
\|q_k\|_{P_c}\le\rho_c^{k/2}\|q_0\|_{P_c}+
\sum_{j=0}^{k-1}\rho_c^{(k-1-j)/2}\|d_j\|_{P_c}.
$$
Si $\|d_j\|_{P_c}\le\bar d$, resulta la cota asintótica $\bar d/(1-\sqrt{\rho_c})$. Es una garantía ISS y puede ser conservadora. El ruido gaussiano no tiene cota determinista uniforme; con segundos momentos uniformemente acotados se obtiene acotación en media cuadrática mediante esta desigualdad y Minkowski. No se promete que cada trayectoria ruidosa converge exactamente a cero.

Para referencia variable $s_{r,k}=[r_k,\dot r_k]^T$, los términos de referencia son
$$
d_k=\begin{bmatrix}
A_d(b_k)s_{r,k}+B_d(b_k)u_{{ff},k}-s_{r,k+1}+w_k\\
A_0s_{r,k}+B_0u_{{ff},k}-s_{r,k+1}+L_o\nu_k
\end{bmatrix}.
$$
Esta expresión permite distinguir el error por referencia/modelo del error por inestabilidad. Para referencias con velocidad constante no acotadas en posición, la estructura $A_d[:,0]=[1,0]^T$ cancela la posición absoluta; importan los incrementos, velocidades y aceleraciones.

```python
reg_safe = simulate('safe')
plot_comparison([reg_naive, reg_lqr, reg_safe],
                'Experimento B · dos propuestas y solución con certificado')
display(pd.DataFrame([metrics(x) for x in [reg_naive, reg_lqr, reg_safe]])
        .set_index('Método').round(4))
print(f'Fracción de ganancias filtradas: {np.mean(reg_safe["theta"] < .999):.1%}')
print(f'Menor margen durante la simulación: {reg_safe["margin"].min():.6g}')
```

Es posible que el filtro no modifique ninguna ganancia del LQR nominal: eso significa que esas propuestas ya pertenecen al conjunto certificado. La garantía existe por la desigualdad matricial, aunque el filtro permanezca inactivo. La diferencia de desempeño frente a EKF+LQR también refleja el cambio de observador; no se debe atribuir exclusivamente al filtro de ganancias.

La referencia oscila suavemente y $b$ cambia de 2 a 4.2 y después a 1 N·s/m. Las tres propuestas reciben el mismo ruido. Ahora hay movimiento continuado que aporta información sobre la fricción, y el paseo aleatorio del EKF permite adaptarse a los saltos. Estos saltos son deliberadamente más rápidos que su modelo medio $b_{k+1}=b_k$.

## Prueba adversa del certificado y activación del filtro

Ahora aislamos la afirmación matemática: referencia constante, sin ruido, y ganancias solicitadas que cambian bruscamente, algunas incluso de signo. **No son salidas observadas del EKF**: son una prueba deliberada de que el certificado no depende de la calidad del proponente. La fricción cambia entre los extremos físicos cada muestra.

Se simula directamente la matriz aumentada exacta y se verifica $V_{k+1}/V_k\le\rho_c$. Esta comprobación numérica ilustra la prueba; la cobertura de todas las fricciones viene del argumento de cajas y convexidad, no de esta única trayectoria.

```python
n_stress = 500
requests = [gain_lqr(2.0), np.array([80., -5.]), np.array([-12., 2.]), gain_lqr(20.)]
filtered = [filter_gain(K) for K in requests]
q = np.array([-2., 0.5, -1.7, -0.3])
qh, vh, ratios, th, applied_hist = [q.copy()], [float(q@P_CERT@q)], [], [], []
for k in range(n_stress):
    j = (k//17) % len(requests)
    K, theta, margin = filtered[j]
    b = B_PHYS[k % 2]
    A, B = model(b)
    q = augmented_matrix(A, B, K) @ q
    vnext = float(q@P_CERT@q)
    ratios.append(vnext/vh[-1]); vh.append(vnext); qh.append(q.copy())
    th.append(theta); applied_hist.append(K)
assert max(ratios) <= RHO_C+1e-10
assert all(item[2] >= TOL_CERT for item in filtered)
stress_t = DT*np.arange(n_stress+1)
fig, ax = plt.subplots(3, 1, figsize=(11, 8), sharex=True)
ax[0].semilogy(stress_t, vh, label='V real')
ax[0].semilogy(stress_t, vh[0]*RHO_C**np.arange(n_stress+1), 'k--', label='Cota ρ^k V₀')
ax[0].set_ylabel('V(q)'); ax[0].legend()
ax[1].plot(stress_t[:-1], ratios); ax[1].axhline(RHO_C, color='k', ls='--')
ax[1].set_ylabel('V siguiente / V actual')
ax[2].step(stress_t[:-1], th, where='post'); ax[2].set_ylabel('θ aceptado')
ax[2].set_xlabel('Tiempo [s]')
fig.suptitle('Experimento D · ganancias arbitrarias, fricción conmutada, mismo P'); fig.tight_layout(); plt.show()
print(f'Mayor cociente observado: {max(ratios):.8f}; límite: {RHO_C}')
display(pd.DataFrame([{'K solicitada': str(k), 'K aplicada': str(a[0]),
                       'θ': a[1], 'Margen': a[2]} for k, a in zip(requests, filtered)]))
```

## Cómo utilizar y ampliar esta solución

**Propuesta práctica final.** Mantén el EKF como estimador de parámetros y generador de propuestas LQR. Certifica el lazo de realimentación completo sobre un intervalo físico y filtra cada ganancia propuesta con el mismo $P_c$. Conserva un respaldo que ya pertenezca al conjunto seguro. Si prefieres realimentar directamente los estados del EKF, presenta la garantía como condicionada a una cota de su error o desarrolla una prueba conjunta específica: no reutilices sin cambios el certificado del observador lineal.

| Cambio de aplicación | Trabajo necesario |
|---|---|
| Nuevo $m$, $T_s$, intervalo de $b$ o ganancias del observador | Reconstruir vértices y resolver/verificar el certificado. |
| Saturación de fuerza | Añadir análisis regional, anti-windup si hay integrador, o MPC robusto con restricciones; recortar $u$ después del filtro invalida esta prueba. |
| Fricción que cambia dentro de una muestra | Obtener una envolvente que cubra esa transición variable; el ZOH con $b_k$ constante no la certifica. |
| Incertidumbre de masa, retardo o dinámica adicional | Ampliar el modelo aumentado y la envolvente de incertidumbre. |
| Sesgo constante o fuerza desconocida | Añadir estimación/integración y volver a analizar estabilidad; la ley actual no garantiza rechazo exacto. |
| Ruido alto o datos atípicos | Revisar covarianzas, innovaciones y observador; una mejora estadística no debe alterar inadvertidamente el lazo certificado. |
| Necesidad de aprender $b$ en reposo | Planificar una excitación compatible con los límites operativos; ningún filtro crea información ausente. |

La evaluación de 48 matrices pequeñas y una eventual bisección es reproducible aquí, pero **no se ha medido tiempo de ejecución de peor caso en hardware de control**. Puede precalcularse una región de ganancias o usar un solver especializado si hay restricciones estrictas de tiempo real.

La garantía es matemática para la familia de modelos declarada, con un certificado obtenido en aritmética flotante y márgenes positivos verificados. No es una prueba formal mediante aritmética de intervalos ni una validación experimental del vehículo. Las simulaciones verifican implementación y permiten comparar desempeño; no sustituyen las hipótesis físicas.
